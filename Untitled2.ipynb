```python
import pandas as pd

CSV_PATH = "/content/data/train_solution_bounding_boxes (1).csv"

df = pd.read_csv(CSV_PATH)

print("Bounding boxes:", len(df))
print("Unique annotated images:", df["image"].nunique())
```

```python
from pathlib import Path
from PIL import Image

TRAIN_IMAGES = Path("/content/data/training_images")
LABELS_DIR = Path("/content/car_labels")

LABELS_DIR.mkdir(parents=True, exist_ok=True)

created = 0
skipped = 0

for image_name, group in df.groupby("image"):

    image_path = TRAIN_IMAGES / image_name

    if not image_path.exists():
        print("Image not found:", image_name)
        skipped += 1
        continue

    # Get image dimensions
    with Image.open(image_path) as image:
        image_width, image_height = image.size

    yolo_labels = []

    # Process every car in this image
    for _, row in group.iterrows():

        xmin = row["xmin"]
        ymin = row["ymin"]
        xmax = row["xmax"]
        ymax = row["ymax"]

        # Convert to YOLO format
        x_center = (xmin + xmax) / 2
        y_center = (ymin + ymax) / 2

        box_width = xmax - xmin
        box_height = ymax - ymin

        # Normalize
        x_center /= image_width
        y_center /= image_height
        box_width /= image_width
        box_height /= image_height

        # Class 0 = car
        yolo_labels.append(
            f"0 {x_center:.6f} {y_center:.6f} "
            f"{box_width:.6f} {box_height:.6f}"
        )

    # One TXT file per image
    label_path = LABELS_DIR / f"{Path(image_name).stem}.txt"

    label_path.write_text("\n".join(yolo_labels))

    created += 1

print("Conversion complete!")
print("Labels created:", created)
print("Images skipped:", skipped)
print("Labels saved to:", LABELS_DIR)
from sklearn.model_selection import train_test_split

annotated_images = df["image"].unique()

train_names, val_names = train_test_split(
    annotated_images,
    test_size=0.20,
    random_state=42
)

print("Training images:", len(train_names))
print("Validation images:", len(val_names))
from pathlib import Path

YOLO_DATASET = Path("/content/car_dataset")

folders = [
    YOLO_DATASET / "images" / "train",
    YOLO_DATASET / "images" / "val",
    YOLO_DATASET / "labels" / "train",
    YOLO_DATASET / "labels" / "val"
]

for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

print("YOLO folder structure created!")


import shutil
from pathlib import Path

SOURCE_IMAGES = Path("/content/data/training_images")
SOURCE_LABELS = Path("/content/car_labels")
YOLO_DATASET = Path("/content/car_dataset")

# Copy training images + labels
for image_name in train_names:

    source_image = SOURCE_IMAGES / image_name
    source_label = SOURCE_LABELS / f"{Path(image_name).stem}.txt"

    destination_image = YOLO_DATASET / "images" / "train" / image_name
    destination_label = YOLO_DATASET / "labels" / "train" / f"{Path(image_name).stem}.txt"

    shutil.copy2(source_image, destination_image)
    shutil.copy2(source_label, destination_label)


# Copy validation images + labels
for image_name in val_names:

    source_image = SOURCE_IMAGES / image_name
    source_label = SOURCE_LABELS / f"{Path(image_name).stem}.txt"

    destination_image = YOLO_DATASET / "images" / "val" / image_name
    destination_label = YOLO_DATASET / "labels" / "val" / f"{Path(image_name).stem}.txt"

    shutil.copy2(source_image, destination_image)
    shutil.copy2(source_label, destination_label)

print("Images and labels copied successfully! ✅")
print("Train images:",
      len(list((YOLO_DATASET / "images" / "train").glob("*.jpg"))))

print("Train labels:",
      len(list((YOLO_DATASET / "labels" / "train").glob("*.txt"))))

print("Validation images:",
      len(list((YOLO_DATASET / "images" / "val").glob("*.jpg"))))

print("Validation labels:",
      len(list((YOLO_DATASET / "labels" / "val").glob("*.txt"))))

from pathlib import Path

yaml_content = """
path: /content/car_dataset

train: images/train
val: images/val

names:
  0: car
"""

yaml_path = YOLO_DATASET / "data.yaml"

yaml_path.write_text(yaml_content)

print("data.yaml created successfully!")
print()
print(yaml_path.read_text())
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")
```

```python
results=model.train(
    data="/content/car_dataset/data.yaml",
    epochs=20,
    imgsz=640,
    batch=16,
    device=0)
```
